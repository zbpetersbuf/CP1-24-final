```python

def rydform(n1,n2):
    Rm = 10973731.6
    return 10**10/(Rm*(1/n2**(2)-1/n1**(2)))

print(rydform(5,2))
```

```python

y_vals =[a1,a2,a3,a4,a5,a6]
y_vals =[1/(y*10**(-10)) for y in y_vals]
x_vals =[8,7,6,5,4,3]
x_vals =[x**(-2) for x in x_vals]


def ry_lin_from(x, R_m, c):
    return R_m * (1/4 - x) + c


popt, _ = curve_fit(ry_lin_from, x_vals, y_vals, p0=[1, 1])

R_m_fit, c_fit = popt
print(f"Linear Fit Equation: y = {calib.round_sf(R_m_fit,3)}(1/4 - x) + {calib.round_sf(c_fit,3)}")
plt.plot(x_vals, y_vals, 'o', label='Data Points')
plt.plot(x_vals, ry_lin_from(np.array(x_vals), R_m_fit, c_fit), label=f'Fit: y = {calib.round_sf(R_m_fit,3)}(1/4 - x) + {calib.round_sf(c_fit,3)}')

plt.xlabel('n_1^(-2)')
plt.ylabel('1/lamda')
plt.title('Rydberg formula, Balmer series')
plt.legend()
plt.grid(True)
plt.show()
```
