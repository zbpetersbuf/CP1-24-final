```python
import importlib
import calib
importlib.reload(calib)
import sys
sys.path.append('/workspaces/CP1-24-final/zbpetersbuf/advlab3data/')  # Adjust if the module path is different

# Set experiment name and filetype
exp_name = "merq"
filetype = ".txt"

# Call the function to read the files
file_contents = calib.read_files(exp_name, filetype)
```

```python
plt.figure(figsize=(10, 6))

# Loop through the file contents to plot each file
for file_path, data in file_contents.items():
    # Extract time (x) and value (y) from the data
    time = [point[0] for point in data]  # Wavelength in Angstroms
    value = [point[1] for point in data]  # Intensity

    # Plot the data for the current file
    plt.plot(time, value, label=file_path.split('/')[-1])  # Extract just the file name for the legend

# Add labels and title
plt.xlabel('Wavelength (Angstroms)')
plt.ylabel('Intensity')
plt.title(f'Intensity vs Wavelength for {exp_name}')

# Add a legend
plt.legend()

# Display the plot
plt.grid(True)
plt.show()
```
