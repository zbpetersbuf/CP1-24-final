```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Given y_vals and x_vals
y_vals = [1/(y*10**(-10)) for y in [a1, a2, a3, a4, a5, a6]]  # Use your values for a1, a2, etc.
x_vals = [x**(-2) for x in [8, 7, 6, 5, 4, 3]]  # As per your data

# Define the linear model (Rydberg equation)
def ry_lin_from(x, R_m, c):
    return R_m * (1/4 - x) + c

# Fit the data to the model
popt, pcov = curve_fit(ry_lin_from, x_vals, y_vals, p0=[1, 1])

# Extract fitted parameters
R_m_fit, c_fit = popt

# Calculate fitted values
y_fit = ry_lin_from(np.array(x_vals), *popt)

# Calculate residuals
residuals = y_vals - y_fit

# Calculate R-squared
ss_res = np.sum(residuals**2)  # Residual sum of squares
ss_tot = np.sum((y_vals - np.mean(y_vals))**2)  # Total sum of squares
r_squared = 1 - (ss_res / ss_tot)

# Calculate Adjusted R-squared
n = len(y_vals)  # Number of data points
p = len(popt)  # Number of parameters (2: R_m and c)
r_squared_adj = 1 - (1 - r_squared) * (n - 1) / (n - p - 1)

# Print fitted parameters and errors
R_m_error = np.sqrt(pcov[0, 0])
c_error = np.sqrt(pcov[1, 1])

print(f"Fitted R_m: {R_m_fit} ± {R_m_error}")
print(f"Fitted c: {c_fit} ± {c_error}")
print(f"R-squared: {r_squared}")
print(f"Adjusted R-squared: {r_squared_adj}")

# Plot the data and the fit
plt.plot(x_vals, y_vals, 'o', label='Data Points')
plt.plot(
    x_vals, y_fit,
    label=(
        f'Fit: y = 9760000(1/4 - x) + 220000\n'
        f'$R^2 = {r_squared:.3f}$, $Adj\\ R^2 = {r_squared_adj:.3f}$'
    )
)




plt.xlabel(r'1 over the square of the quantum number $n_1^{-2}$')  # Label for x-axis
plt.ylabel(r'Inverse Wavelength, $\lambda^{-1}$ [1/$\mathrm{\AA}$]')  # Label for y-axis

# Title and legend
plt.title('Rydberg formula, for the Balmer series')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```
