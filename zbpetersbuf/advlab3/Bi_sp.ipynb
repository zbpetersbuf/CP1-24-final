```python
import importlib
import absospec
importlib.reload(absospec)
import sys
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import savgol_filter

sys.path.append('/workspaces/CP1-24-final/zbpetersbuf/advlab3data/')

file_contents = absospec.read_files('I2_wiggle.txt')

plt.figure(figsize=(10, 6))

for file_path, data in file_contents.items():
    wavelength = np.array([0.98*point[0] + 35.5 for point in data])
    intensity = np.array([point[1] for point in data])

    # Apply Savitzky-Golay filter
    smooth_intensity = savgol_filter(intensity, window_length=wl, polyorder=pol)

    # Detrended data = original - smoothed
    detrended_intensity = intensity - smooth_intensity

    # Plot detrended data
    plt.plot(wavelength, detrended_intensity, label='Detrended Signal')

plt.xlabel('Wavelength (Angstroms)')
plt.ylabel('Detrended Intensity')
plt.title('Detrended Spectrum (Original - Smoothed)')

#plt.xlim(3500, 7000)
plt.xlim(xmin, xmax)
plt.legend()
plt.grid(True)
plt.show()
```

```python
import importlib
import Birge_Sponer
importlib.reload(Birge_Sponer)
import calib
importlib.reload(calib)
import sys

x = 10

v = np.arange(x, x + 36)

# Call your Birge-Sponer function to get parameters
w_e, x_e = Birge_Sponer.Birge_Sponer_funk(v)


lamdas = [5172, 5186, 5200, 5215, 5230, 5245, 5261, 5278, 5296, 5314, 5333, 5352, 5373, 5394, 5415, 5439, 5464, 5490,
          5515, 5541, 5568, 5595, 5623, 5651, 5682, 5712, 5743, 5775, 5807, 5840, 5873, 5908, 5943, 5980, 6015, 6052]



G_v = [1 / (x * 1e-8) for x in lamdas]

# Define fit function (unless already imported)
def fit_BS(v, w_e, x_e):
    return w_e - 2 * w_e * x_e * (v + 1)

# Generate fit for plotting
v_fit = np.linspace(min(v), max(v), 200)
G_fit = fit_BS(v_fit, w_e, x_e)

# Plot
plt.figure(figsize=(8, 5))
plt.scatter(v, G_v, label='Data', color='blue')
plt.plot(v_fit, G_fit, label='Birge-Sponer Fit', color='orange')
plt.xlabel('Vibrational Quantum Number v')
plt.ylabel('1 / λ (cm⁻¹)')
plt.title('Birge-Sponer Fit')
plt.legend()
plt.grid(True)
plt.show()

# Print results
print(f"ω_e = {w_e:.2f}, x_e = {x_e:.4f}")
```
